Accept 5 paper marks, roll no.,name; and calculate total and percentage if percentage >= 85 , student can take admission else can't take admission using class and object and constructor

```python
class Student:
    def __init__(self):
        self.roll_no = int(input("Enter Roll Number: "))
        self.name = input("Enter Student Name: ")

        self.marks = []
        for i in range(5):
            mark = float(input(f"Enter marks of subject {i+1}: "))
            self.marks.append(mark)

    def calculate(self):
        self.total = sum(self.marks)
        self.percentage = self.total / 5

    def display(self):
        print("\n----- Student Result -----")
        print("Roll Number:", self.roll_no)
        print("Student Name:", self.name)
        print("Total Marks:", self.total)
        print("Percentage:", self.percentage, "%")

        if self.percentage >= 85:
            print("Eligible for Admission")
        else:
            print("Not Eligible for Admission")


s1 = Student()
s1.calculate()
s1.display()
```

```python
# Single Inheritence
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

obj=B()
obj.showA()
obj.showB()
```

```python
# Multiple child
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")
class C(B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
#Single child
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")
class C(B, A):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
# Hybrid Inheritance
class ParentA:
  def showA(self):
    print("showA function")

class ChildB(ParentA):
  def showB(self):
    print("showB function")

class ChildC(ParentA):
  def showC(self):
    print("showC function")

class HybridD(ChildB, ChildC):
  def showD(self):
    print("showD function")

obj = HybridD()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

```python
# Hierarchical Inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C(A):
  def showC(self):
    print("showC function")

class D(A):
  def showD(self):
    print("showD function")

objB = B()
objC = C()
objD = D()
objB.showA()
objB.showB()
objC.showA()
objC.showC()
objD.showA()
objD.showD()
```
