### Task:

Program to print the area of a rectangle by creating a class named 'Area' taking the values of its length and breadth as parameters of its constructor and having a method named 'returnarea' which returns the area of the rectangle. length and breadth of rectangle are entered through keyboard

```python
class Area:
  def __init__(self,l,b):
    self.length=l
    self.breadth=b
  def returnArea(self):
    return self.length*self.breadth

length=int(input("Enter the length: "))
breadth=int(input("Enter the breadth: "))
obj=Area(length,breadth)
print("Area of Rectangle is", obj.returnArea())
```

If the child class is unable to provide implementation of all methods of abstrat class then we can declare the child class methods with @abstractmethod and complete the remaining method implementation

```python
from abc import ABC , abstractmethod
#from abc import *
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

  @abstractmethod
  def m2(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implemetation")
  def m2(self):
    pass
class Child(MyPython):
  def m2(self):
    print("m2 implementation")
m=Child()
m.m1()
m.m2()
```

```python
# We can use seek() function to move cursor(file pointer) to
# specified location.

data = "Learning python is very easy from Ashish Sir"
f = open("myfile.txt", 'w')
f.write(data)

with open("myfile.txt", 'r+') as f:
    fileData = f.read()
    print(fileData)
    print("Current Cursor Position is ", f.tell())
    f.seek(15)
    print("Current Cursor Position is ", f.tell())
    f.write("-JAVA-")
    f.seek(0)
    fileData = f.read()
    print(fileData)
```

```python
# WAP to print number of lines, words, characters present in file

import os, sys

fname = input("Enter file name : ")

if os.path.isfile(fname):
    print("File exist : ", fname)
    f = open(fname, 'r')
else:
    print("File is not exist : ", fname)
    sys.exit(0)

lcount = wcount = ccount = 0

for line in f:
    lcount += 1
    ccount = ccount + len(line)
    words = line.split()
    wcount = wcount + len(words)

print("Total lines : ", lcount)
print("Total words : ", wcount)
print("Total characters : ", ccount)
```
