### NumPy:

```python
import numpy as np
```

```python
students = np.array([
    [85,90,78],
    [60,55,70],
    [95,92,88],
    [40,50,45],
    [72,68,75]
])
students
```

```python
avg_marks = np.mean(students, axis=1)
print("Average Marks:",avg_marks)
```

```python
avg_marks = np.mean(students, axis=0)
print("Average Marks:",avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("Topper Student:",topper)
```

```python
failed_student = np.where(avg_marks<90)
print("Failed Student:",failed_student[0]+1)
```

```python
for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 50 else "Fail"
    print(f"Student = {i+1}:Avg = {avg:.2f} {status}")
```

### Matplotlib:

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,60,8,100]
plt.plot(x,y)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,60,8,150]
plt.plot(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,60,8,100]
plt.scatter(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,4,60,8,100]
plt.bar(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000, 3200, 3500, 3800]
plt.bar(months, sales, color='maroon')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
plt.barh(months, sales, color='forestgreen')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
height = [150, 160,170, 180, 190]
weight = [60, 65, 70, 75, 80]
plt.scatter(height, weight)
plt.title('Height vs Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.grid(True)
plt.show()
```

```python
brands = ['A', 'B', 'C', 'D']
sales = [100, 200, 150, 300]
plt.pie(sales, labels=brands, autopct='%1.1f%%',startangle = 90,explode=(0,0.1,0,0))
plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [2,4,60,8,100]
y2 = [1,3,5,7,9]
plt.plot(x,y1,label='Line 1')
plt.plot(x,y2,label='Line 2')
plt.title('Multiple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
```

```python
# Subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# 1. Line Plot
axs[0, 0].plot(x, y)
axs[0, 0].set_title('Square')

# 2. Bar Plot (using original 12-month sales data)
monthly_sales = [1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000, 3200, 3500, 3800]
axs[0, 1].bar(months, monthly_sales)
axs[0, 1].set_title('Sales')
axs[0, 1].tick_params(axis='x', rotation=45)

# 3. Scatter Plot
axs[1, 0].scatter(height, weight)
axs[1, 0].set_title('Scatter')

# 4. Pie Chart (using the 4-element sales data)
pie_sales = [100, 200, 150, 300]
axs[1, 1].pie(pie_sales, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Brands')

plt.tight_layout()
plt.show()
```
