```python
# Writing data to .CSV file

import csv

with open("emp.csv", "w", newline="") as f:
    w = csv.writer(f)

    w.writerow(["ID", "NAME", "SALARY", "ADDRESS"])

    n = int(input("Enter number of employees: "))

    for i in range(n):
        eno = input("Enter employee no: ")
        ename = input("Enter employee name: ")
        esal = input("Enter employee salary: ")
        eaddr = input("Enter employee address: ")

        w.writerow([eno, ename, esal, eaddr])

print("Data inserted successfully!")
```

```python
import csv

f = open("emp.csv", "r")
r = csv.reader(f)
data = list(r)
#print(data)

for line in data:
    for word in line:
        print(word, '\t', end='')

    print()
```

```python
# Zip operation

from zipfile import *
f = ZipFile("myzip.zip", 'w', ZIP_DEFLATED)
f.write("/content/women.jpg")
f.write("/content/youngwoman.jpg")
f.write("/content/emp.csv")
f.write("/content/file1.py")
f.close()
print("myzip.zip is created....")
```
